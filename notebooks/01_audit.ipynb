```python
from pathlib import Path
import pandas as pd

RAW = Path("../data/raw")

files = sorted(p for p in RAW.rglob("*") if p.is_file())

inventory = pd.DataFrame({
    "file": [str(p.relative_to(RAW)) for p in files],
    "type": [p.suffix.lower() for p in files],
    "size_MB": [round(p.stat().st_size / 1e6, 2) for p in files],
})

print(f"{len(inventory)} files, {inventory.size_MB.sum():.1f} MB total")
inventory
```

```python
sample = files[0]

print("Reading:", sample)
print("\nFirst 8 lines:\n")

with open(sample, "r", errors="replace") as f:
    for _ in range(8):
        print(f.readline().rstrip())
```

```python
df = pd.read_csv(sample, nrows=200_000)  
display(df.head())
df.info()
display(df.describe(include="all").T)
```

```python
print("Rows:", len(df))
print("\nMissing values per column:")
print(df.isna().sum())
print("\nDuplicate rows:", df.duplicated().sum())
print("\nColumn dtypes:")
print(df.dtypes)
```

```python
import pandas as pd

audit = pd.read_csv("../reports/audit_summary.csv", parse_dates=["start", "end"])
outage = pd.read_csv("../reports/outage_summary.csv")
events = pd.read_csv("../reports/outage_events.csv", parse_dates=["start", "end"])

quality = audit[["house_id", "start", "end", "days_span", "repeated_value_pct",
                 "mean_kW", "max_kW"]].merge(outage, on="house_id")
quality.sort_values("zero_pct", ascending=False)
```

```python
import sys; sys.path.append("..")
import numpy as np, pandas as pd, plotly.express as px
from src.data.loading import load_house

reg = pd.read_csv("../reports/house_registry.csv", parse_dates=["start", "end"])
print(reg.group.value_counts())

# 4 development houses spanning low to high consumption, each with 3+ sub-meters
cand = reg[(reg.group == "model") & (reg.n_submeters >= 3)].sort_values("mean_kW")
picks = np.linspace(0, len(cand) - 1, 4).round().astype(int)
cand.iloc[picks][["house_id", "mean_kW", "max_kW", "n_submeters", "frozen_pct"]]
```

```python
events = pd.read_csv("../reports/outage_events.csv", parse_dates=["start", "end"])
long_zero = events[(events.kind == "zero") & (events.minutes >= 360)].copy()
long_zero["house"] = "House " + long_zero.house_id.astype(str)
print(len(long_zero), "zero stretches of 6+ hours in", long_zero.house_id.nunique(), "houses")

fig = px.timeline(long_zero, x_start="start", x_end="end", y="house", color="house",
                  title="Zero-usage stretches of 6+ hours")
fig.update_yaxes(autorange="reversed")
fig.show()
```

```python
h2 = load_house("../data/raw/precon/House2.csv")
zero_share = (h2["Usage_kW"] == 0).astype(float).resample("D").mean()
bad_days = zero_share[zero_share >= 0.95].index
print("Fully-zero days:", len(bad_days), "| first one:", bad_days[0].date())

day = h2.loc[str(bad_days[0].date()), "Usage_kW"]
print(day.describe())
print("\nNon-zero readings on that day:")
print(day[day > 0])
```
