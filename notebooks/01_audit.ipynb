```python
from pathlib import Path
import pandas as pd

RAW = Path("../data/raw")

files = sorted(p for p in RAW.rglob("*") if p.is_file())

inventory = pd.DataFrame({
    "file": [str(p.relative_to(RAW)) for p in files],
    "type": [p.suffix.lower() for p in files],
    "size_MB": [round(p.stat().st_size / 1e6, 2) for p in files],
})

print(f"{len(inventory)} files, {inventory.size_MB.sum():.1f} MB total")
inventory
```

```python
sample = files[0]

print("Reading:", sample)
print("\nFirst 8 lines:\n")

with open(sample, "r", errors="replace") as f:
    for _ in range(8):
        print(f.readline().rstrip())
```

```python
df = pd.read_csv(sample, nrows=200_000)  
display(df.head())
df.info()
display(df.describe(include="all").T)
```

```python
print("Rows:", len(df))
print("\nMissing values per column:")
print(df.isna().sum())
print("\nDuplicate rows:", df.duplicated().sum())
print("\nColumn dtypes:")
print(df.dtypes)
```

```python
import pandas as pd
import plotly.express as px

summary = pd.read_csv("../reports/audit_summary.csv", parse_dates=["start", "end"])
gaps = pd.read_csv("../reports/audit_gaps.csv", parse_dates=["last_seen_before", "next_seen_after"])

summary.sort_values("coverage_pct")
```

```python
print(summary["columns"].value_counts())
```

```python
fig = px.timeline(summary, x_start="start", x_end="end", y="house_id",
                  color="coverage_pct", title="Recording period per house")
fig.update_yaxes(type="category", autorange="reversed")
fig.show()
```
